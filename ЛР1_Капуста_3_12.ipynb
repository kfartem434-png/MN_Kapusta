### Завдання 25
Дано два вектори:

```text
a = [1, 2]
b = [2, 1]
```

Обчисліть косинус кута між ними за формулою через скалярний добуток і норми векторів.

Додатково знайдіть сам кут у градусах.

````python
# Код студента
a = np.array([1, 2])
b = np.array([2, 1])
a_norm = np.linalg.norm(a)
b_norm = np.linalg.norm(b)
vcosine = (np.dot(a,b))/(a_norm * b_norm)

print("Косинус кута між векторами: \n",vcosine)
print("Кут між векторами у градусах: \n",np.degrees(np.arccos(vcosine)))
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array(
    [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
)
mean = np.mean(A,axis=0)
std = np.std(A,axis=0)

normalized = (A - mean)/std

print("Нормалізовані стовпці: \n",normalized)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array(
    [[1, 2, 1],
     [2, 1,-1],
     [3,-1, 2]]
)
b = np.array([9,8,3])

solution = np.linalg.solve(A,b)

print("Матричне множення: \n",A @ solution)
print("Порівняння з вектором b: \n",np.allclose(A @ solution,b))
````
